```python
# 1. 
import pandas as pd
from sklearn import tree
from sklearn.metrics import ConfusionMatrixDisplay, classification_report
import seaborn as sns
from sklearn.preprocessing import StandardScaler
diabetes_df= pd.read_csv('diabetes.csv')
diabetes_df.head()
```

```python
from sklearn.model_selection import train_test_split

X= diabetes_df.drop(['Outcome'], axis=1)
y= diabetes_df['Outcome']
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.3, random_state=40)

```

```python

sc = StandardScaler()
X_train = sc.fit_transform(X_train)
X_test = sc.fit_transform(X_test)

```

```python
model = tree.DecisionTreeClassifier(max_depth=5, random_state=40)
```

```python
model = model.fit(X_train, y_train)
y_pred = model.predict(X_test)
```

```python
ConfusionMatrixDisplay.from_predictions(y_test, y_pred)
```

```python
# 2. Create a function that accepts an array of names and returns a string formatted as a list of names separated by commas EXCEPT for the last two names, which are separated by an ampersand (and sign - &)
def order_names(names):
    if len(names) == 0:
        return ""
    elif len(names) == 1:
        return names[0]
    elif len(names) == 2:
        return f"{names[0]} & {names[1]}"
    else:
        return ", ".join(names[:-1]) + f" & {names[-1]}"

order_names(['Andrew','John', 'George', 'Sam'])
```
