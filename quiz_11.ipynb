```python
# 1. Create a function that accepts a list of values. Filter out all values that are numeric and return the new list.
def letters (char_list):
    """
    We want to create a function that filters out all numeric values
    """
    filtered_list = [x for x in char_list if isinstance(x, str)]
    return filtered_list

letters([1,2,'a','b',7.6])




```

```python
#2. Write a clean solution to fizzbuzz that we did in a previous class that uses default arguments, typing, a docstring, and useful comments
def fizzbuzz(limit: int = 100):
    """
    This function takes in a number up to 100 and then 
    for every number up to the input returns the number
    and fizz if it is divisble by 3, buzz if is divisble by 5
    and fizzbuzz if it is divisible by both 3 and 5
    """
    for i in range(1,limit+1):
        if i % 3 == 0 and i % 5 == 0:
            print(i, "fizzbuzz")
        elif i % 3 == 0:
            print(i,"fizz")
        elif i % 5 == 0:
            print(i, "buzz")
        else:
            print(i)

fizzbuzz(30)

```
